```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.arange(0, 1.001, 0.001)

f = 5

signal = 3 * np.sin(2 * np.pi * f * t)
```

```python
plt.plot(signal)
plt.title('original signal')
plt.grid(True)
plt.show()
```

```python
#Sampling different frequencies
samplingFrequencies = [15, 20, 25, 50, 100]; 

for i in range(len(samplingFrequencies)):
    fs = samplingFrequencies[i]
    ts = np.arange(0, 1 + (1/fs), 1/fs)
    sampledSignal = np.sin(2 * np.pi * f * ts)
    plt.subplot(3, 2, i + 2)

    plt.plot(t,signal)
    plt.stem(ts, sampledSignal)
    plt.grid(True)
    plt.title(f"sampled at: {fs}Hz")
```

```python
# sampling multiple frequency component signal 
s2 = 2 * np.sin(2 * np.pi * 10 * t)
s3 =  np.sin(2 * np.pi * 20 * t)
ms = signal + s2 + s3
#Highest frequency component = 20
fsample = 50 #Hz
tsample = np.arange(0, 1 + (1/fsample), 1/fsample)

sampledSignal = np.sin(2 * np.pi * f * tsample) + np.sin(2 * np.pi * 20 * tsample) + (2 * np.sin(2 * np.pi * 10 * tsample))

plt.plot(t,ms)
plt.stem(tsample, sampledSignal)
plt.grid(True)
plt.title(f"multiple frequency signal sampled at: {fsample}Hz")
```
